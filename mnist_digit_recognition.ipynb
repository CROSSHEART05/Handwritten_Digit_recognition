## 📐 Mathematical Formulation

### 1. Convolutional Layer Operations
For an input feature map $X$ and kernel $W$, convolution at pixel position $(i, j)$ is given by:

$$S(i, j) = (X * W)(i, j) = \sum_{m} \sum_{n} X(i - m, j - n) W(m, n) + b$$

Followed by Batch Normalization and ReLU non-linearity:

$$\text{ReLU}(x) = \max(0, x)$$

### 2. Multi-Digit Contour Extraction & Spatial Normalization
Given image $I(x, y)$, binary thresholding isolates digit regions $C_k$. Spatial moments are calculated to center each digit's mass at $(14, 14)$ inside a $28 \times 28$ matrix:

$$\bar{x} = \frac{M_{10}}{M_{00}}, \quad \bar{y} = \frac{M_{01}}{M_{00}} \quad \text{where } M_{pq} = \sum_x \sum_y x^p y^q I(x, y)$$

```python
# Step 1: Environment Setup & Library Imports
import os
import io
import base64
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
import torch
import torch.nn as nn
import torch.nn.functional as F
from IPython.display import HTML, display, Javascript

# Import custom training and inference module
from train_model import MNISTNet, segment_and_predict_digits, get_device, train_and_save_model

device = get_device()
print(f"[+] System initialized on compute device: {device}")
```

## 🧠 Step 2: Neural Network Architecture & Model Loading

```python
# Load pre-trained PyTorch CNN Model
model = MNISTNet().to(device)
model_path = "mnist_cnn.pth"

if os.path.exists(model_path):
    print(f"[+] Loading model checkpoint from {model_path}...")
    model.load_state_dict(torch.load(model_path, map_location=device))
else:
    print(f"[*] Model checkpoint not found. Training model...")
    model = train_and_save_model(model_save_path=model_path, epochs=2)

model.eval()
print("[+] PyTorch CNN Model successfully loaded and ready for inference!")
```

## 🔍 Step 3: Computer Vision Multi-Digit Segmentation Visualization
Let's test the computer vision contour extraction algorithm on a generated multi-digit test image (e.g. digits `4`, `2`, `7` side-by-side).

```python
# Create synthetic multi-digit canvas image for CV testing
test_canvas = np.zeros((200, 600), dtype=np.uint8)

# Draw digits 5, 8, 3 using OpenCV font/lines
cv2.putText(test_canvas, "5", (50, 150), cv2.FONT_HERSHEY_SIMPLEX, 4.5, 255, 12)
cv2.putText(test_canvas, "8", (240, 150), cv2.FONT_HERSHEY_SIMPLEX, 4.5, 255, 12)
cv2.putText(test_canvas, "3", (430, 150), cv2.FONT_HERSHEY_SIMPLEX, 4.5, 255, 12)

# Run segmentation and prediction
digit_seq, details, annotated_img = segment_and_predict_digits(test_canvas, model, device)

print(f"[✓] Detected Multi-Digit Sequence: '{digit_seq}'")

# Plot original, thresholded contours & per-digit crops
fig, axes = plt.subplots(1, 1 + len(details), figsize=(14, 4))
axes[0].imshow(cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB))
axes[0].set_title(f"Detected Bounding Boxes: '{digit_seq}'", fontsize=12, fontweight='bold')
axes[0].axis('off')

for i, d in enumerate(details):
    crop_28 = np.array(d['processed_28x28'])
    axes[i+1].imshow(crop_28, cmap='gray')
    axes[i+1].set_title(f"Digit #{d['digit_index']}: '{d['predicted_digit']}' ({d['confidence']}%")
    axes[i+1].axis('off')

plt.tight_layout()
plt.show()
```

## 🌐 Step 5: Gradio Interactive Web Application (Notebook Embedded)
Launch an interactive Gradio interface directly inside the notebook cell for quick live testing.

```python
import gradio as gr

def gradio_predict(sketch_image):
    if sketch_image is None:
        return "Ready", "Draw digits on canvas above."
    
    full_str, details, annotated_img = segment_and_predict_digits(sketch_image, model, device)
    if not full_str:
        return "No Digit Detected", "Please write digits clearly."
    
    breakdown_txt = "\n".join([
        f"Digit #{d['digit_index']}: '{d['predicted_digit']}' | Confidence: {d['confidence']}% | Top-3: {d['top3']}"
        for d in details
    ])
    
    return f"Detected Sequence: '{full_str}'", breakdown_txt

with gr.Blocks(theme=gr.themes.Soft(primary_hue="cyan", dark_mode=True)) as demo:
    gr.Markdown("# ⚡ Gradio Multi-Digit MNIST Recognizer")
    with gr.Row():
        with gr.Column():
            canvas_input = gr.Sketchpad(shape=(600, 200), label="Handwriting Canvas")
            btn = gr.Button("✨ Run PyTorch Inference", variant="primary")
        with gr.Column():
            output_seq = gr.Textbox(label="Detected Number Sequence", placeholder="Results appear here...")
            output_details = gr.Textbox(label="Per-Digit Breakdown", lines=6)
            
    btn.click(fn=gradio_predict, inputs=canvas_input, outputs=[output_seq, output_details])

# Launch Gradio app in inline mode
demo.launch(inline=True, share=False, height=450)
```
